```python

import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Upload image
uploaded = files.upload()
filename = next(iter(uploaded))

# Read image
img = cv2.imread(filename)

# Split RGB channels
b, g, r = cv2.split(img)

# Display original image
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

# Red histogram
plt.subplot(2, 2, 2)
plt.hist(r.ravel(), 256, [0, 256], color='red')
plt.title("Red Histogram")

# Green histogram
plt.subplot(2, 2, 3)
plt.hist(g.ravel(), 256, [0, 256], color='green')
plt.title("Green Histogram")

# Blue histogram
plt.subplot(2, 2, 4)
plt.hist(b.ravel(), 256, [0, 256], color='blue')
plt.title("Blue Histogram")

plt.tight_layout()
plt.show()

print("RGB Histogram Analysis Completed")
```

```python

import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Upload image
uploaded = files.upload()
filename = next(iter(uploaded))

# Read image
img = cv2.imread(filename)

# Convert to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Apply Gaussian Blur
blur = cv2.GaussianBlur(gray, (5, 5), 0)

# Apply Canny Edge Detection
edges = cv2.Canny(blur, 100, 200)

# Display original image
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

# Display edge image
plt.subplot(1, 2, 2)
plt.imshow(edges, cmap='gray')
plt.title("Canny Edge Detection")
plt.axis("off")

plt.show()

# Save output
cv2.imwrite("edge_output.jpg", edges)

print("Edge Detection Completed Successfully")
```

```python

import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Upload image
uploaded = files.upload()
filename = next(iter(uploaded))

# Read grayscale image
img = cv2.imread(filename, 0)

# Create structuring element
kernel = np.ones((5, 5), np.uint8)

# Apply dilation
dilated = cv2.dilate(img, kernel, iterations=1)

# Display images
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(dilated, cmap='gray')
plt.title("Dilated Image")
plt.axis("off")

plt.show()

# Save output
cv2.imwrite("dilated_image.jpg", dilated)

print("Dilation Completed Successfully")
```

```python

import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Upload image
uploaded = files.upload()
filename = next(iter(uploaded))

# Read image
img = cv2.imread(filename)

rows, cols = img.shape[:2]

# Original control points
pts1 = np.float32([
    [50, 50],
    [200, 50],
    [50, 200]
])

# Destination control points
pts2 = np.float32([
    [100, 100],
    [250, 100],
    [100, 250]
])

# Calculate transformation matrix
matrix = cv2.getAffineTransform(pts1, pts2)

print("Transformation Matrix:")
print(matrix)

# Apply affine transformation
result = cv2.warpAffine(img, matrix, (cols, rows))

# Display images
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(result, cv2.COLOR_BGR2RGB))
plt.title("Transformed Image")
plt.axis("off")

plt.show()

# Save output
cv2.imwrite("affine_output.jpg", result)

print("Affine Transformation Completed")
```
